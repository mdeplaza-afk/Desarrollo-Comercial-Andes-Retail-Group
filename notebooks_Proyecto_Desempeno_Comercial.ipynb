### Narrativa del Dashboard (Modelo SCQA)
Se describe de forma estructurada la historia completa del dashboard usando SCQA, incluyendo contexto, problemas detectados, preguntas de negocio y recomendaciones basadas en datos.

#### 🖥️ Vista General (Overview)

**S (Situación):**  
S – Situación (Contexto del negocio)
La compañía ha mantenido un desempeño comercial estable entre 2024 y 2025, con un portafolio diversificado y presencia en múltiples regiones y canales. Sin embargo, al analizar la evolución mensual de ingresos, márgenes y unidades vendidas, se observa un patrón recurrente: las ventas no siguen una tendencia lineal, sino que presentan picos y valles pronunciados a lo largo del año.
Este comportamiento se repite en la mayoría de categorías y canales, sobretodo durante el año 2025 que muestra disminución en el crecimiento con respecto al año 2024, lo que sugiere que existe un factor transversal que condiciona el rendimiento 

**C (Complicación):**  
C – Complicación (Problemas detectados)
El análisis detallado revela que la estacionalidad es el factor que más afecta las ventas, generando meses de alta demanda y otros de contracción significativa.
Esto trae varias complicaciones:
Las comparaciones intermensuales pueden ser engañosas si no se ajustan por estacionalidad.
Algunos productos muestran alta sensibilidad al ciclo estacional, generando riesgos de inventario y rotación.
La rentabilidad depende en gran medida de los meses pico; fuera de ellos, los márgenes se reducen.
Las metas comerciales y presupuestos no están alineados con el comportamiento real del ciclo anual.
En resumen: sin incorporar la estacionalidad, las decisiones comerciales pueden ser imprecisas o reactivas.

**Q (Pregunta):**   
Q – Pregunta (Preguntas de negocio que el dashboard responde)
El dashboard se construye para responder preguntas críticas como:
¿Cómo cambia el desempeño comercial cuando se incorpora la estacionalidad al análisis?
¿Qué meses representan oportunidades de crecimiento y cuáles requieren estrategias defensivas?
¿Qué categorías, regiones y canales son más sensibles al ciclo estacional?
¿Cómo debe ajustarse la planificación de inventarios, campañas y metas según el calendario anual?
¿Qué segmentos sostienen la rentabilidad en meses valle y cuáles dependen exclusivamente de los picos?

Estas preguntas permiten pasar de una lectura superficial del desempeño a una gestión estratégica basada en patrones reales del mercado.

**A (Respuesta):**  
A – Respuesta (Hallazgos y recomendaciones basadas en datos)
Hallazgos clave del Overview
La estacionalidad explica la mayor parte de la variación mensual en ingresos y unidades.
Los meses pico concentran un porcentaje desproporcionado del total anual.
La rentabilidad mejora significativamente en los meses de alta demanda.
La caída en meses valle no se debe a problemas de producto o canal, sino al ciclo natural del mercado.

Hallazgos clave del Detalle
Algunas categorías son altamente estacionales (picos muy marcados), mientras otras mantienen estabilidad.
Ciertos clientes corporativos suavizan la estacionalidad, actuando como “amortiguadores” del ciclo.
Existen productos con buen volumen pero baja rentabilidad en meses valle, y dependiendo del segmento de clientes, lo que requiere ajustes tácticos.


#### 🔎 Vista Detalle

**S (Situación):**  
S – Situación (Contexto del análisis detallado)
La Vista Detalle consolida toda la información transaccional en una tabla única que permite explorar el comportamiento comercial a nivel granular. **C (Complicación):**  
C – Complicación (Problemas detectados al profundizar)
Al profundizar en los datos, se identifican patrones que no son visibles en la Vista General. En particular:
La estacionalidad afecta de manera distinta a cada categoría y los segmentos de clientes.
Algunos productos muestran alta dependencia de los meses pico, mientras otros mantienen estabilidad.
Existen clientes y regiones que amplifican la estacionalidad, generando picos muy altos y valles muy profundos.
En meses valle, ciertos productos mantienen volumen pero pierden margen, lo que afecta la rentabilidad.
La dispersión entre segmentos revela que no todos los comportamientos responden al mismo ciclo estacional.
Estos hallazgos muestran que la estacionalidad no impacta de forma homogénea y que la gestión debe ser diferenciada por segmento.
